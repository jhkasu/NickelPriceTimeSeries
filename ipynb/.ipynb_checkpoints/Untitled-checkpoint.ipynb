```python
# 1. Print it revser order
word = ["apple", "banana", "cherry", "date"]

reverse_words = word[::-1]

print(reverse_words)

# Print each word backword
res = []
for w in word:
    res.append(w[::-1])
print(res)
```

```python
# 3. Having the folloing dic.
# score = {"Ali" : 90, "Sara" : 85, "John": 92, "Lina" : 88}
# Sort the dic by score(value) in descending order
# Print each name and score in order

scores = {"Ali" : 90, "Sara" : 85, "John": 92, "Lina" : 88}
sorted_score = list(scores.values())
sorted_score.sort(reverse=True)
print(sorted_score)
new_dict = {}
for value in sorted_score:
    for name, score in scores.items():
        if score == value:
            new_dict[name] = score
print(new_dict)
```
